```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = yf.download(
    "RELIANCE.NS",
    start="2020-01-01",
    end="2025-01-01"
)

df.head()
```

```python
plt.figure(figsize=(12,6))

plt.plot(df['Close'])

plt.title("Reliance Stock Closing Price")
plt.xlabel("Date")
plt.ylabel("Price")

plt.show()
```

```python
df.to_csv("../data/raw/reliance.csv")
```
